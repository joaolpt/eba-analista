## Ex 1
Crie uma classe Cliente que tenha os atributos cliente_id, nome, idade e cidade, e um método para exibir as informações do cliente.

```python
class Cliente():
    def __init__(self, cliente_id, nome, idade, cidade):
        self.cliente_id = cliente_id
        self.nome = nome
        self.idade = idade
        self.cidade = cidade
    
    def imprimir_info(self):
        print(f"Cliente ID: {self.cliente_id} | Nome: {self.nome} | Idade: {self.idade} | Cidade: {self.cidade}")

cliente1 = Cliente(1, "João Luís", 22, "Bambuí-MG")

cliente1.imprimir_info()
```

## Ex 2
Adicione um método na classe Cliente chamado realizar_compra, que recebe um produto e um valor e exibe uma mensagem confirmando a compra. 

```python
class Cliente():
    def __init__(self, cliente_id, nome, idade, cidade):
        self.cliente_id = cliente_id
        self.nome = nome
        self.idade = idade
        self.cidade = cidade
    
    def imprimir_info(self):
        print(f"Cliente ID: {self.cliente_id} | Nome: {self.nome} | Idade: {self.idade} | Cidade: {self.cidade}")
    
    def realizar_compra(self, produto, valor):
        print(f"Compra Confirmada! {self.nome} sua compra de {produto} no valor de R${valor}, foi efetuada!")

cliente1 = Cliente(1, "João Luís", 22, "Bambuí-MG")
cliente1.realizar_compra("Notebook", 4500.00)
```

## Ex 3
Crie uma classe Compra com atributos compra_id, cliente_id, produto, valor e data_compra, e um método detalhes_compra() para exibir as informações.

```python
import datetime
agora =  datetime.datetime.now().strftime("%d/%m/%Y")
class Compra():
    def __init__(self, compra_id, cliente_id, produto, valor, data_compra):
        self.compra_id = compra_id
        self.cliente_id = cliente_id
        self.produto = produto
        self.valor = valor
        self.data_compra = data_compra

    def detalhes_compra(self):
        print(f"ID da Compra: {self.compra_id} | ID do Cliente: {self.cliente_id} | Produto: {self.produto} | Valor: R${self.valor} | Data: {self.data_compra}")

carrinho1 = Compra(1, 3, "Notebook", 4500.00, agora)

carrinho1.detalhes_compra()
```

## Ex 4
Implemente um método listar_compras() na classe Cliente, que recebe uma lista de objetos Compra e imprime todas as compras feitas pelo cliente.

```python
import datetime
agora =  datetime.datetime.now().strftime("%d/%m/%Y")
class Compra():
    def __init__(self, compra_id, cliente_id, produto, valor, data_compra):
        self.compra_id = compra_id
        self.cliente_id = cliente_id
        self.produto = produto
        self.valor = valor
        self.data_compra = data_compra

    def detalhes_compra(self):
        print(f"ID da Compra: {self.compra_id} | ID do Cliente: {self.cliente_id} | Produto: {self.produto} | Valor: R${self.valor} | Data: {self.data_compra}")

    def listar_compras(self, lista):
        for items in lista:
            items.detalhes_compra()

compra1 = Compra(1, 3, "Notebook", 4500.00, agora)
compra2 = Compra(2, 3, "Mouse", 200.00, agora)
compra3 = Compra(3, 3, "Teclado", 500.00, agora)
carrinho = [compra1, compra2, compra3]

compra1.listar_compras(carrinho)
```

## Ex 5
Crie uma classe Pagamento que armazena pagamento_id, cliente_id, valor_pago, metodo_pagamento e data_pagamento. Inclua um método para exibir os detalhes do pagamento.

```python
import datetime
agora = datetime.datetime.now().strftime("%d/%m/%Y")
class Pagamento():
    def __init__(self, pagamento_id, cliente_id, valor_pago, metodo_pagamento, data_pagamento):
        self.pagamento_id = pagamento_id
        self.cliente_id = cliente_id
        self.valor_pago = valor_pago
        self.metodo_pagamento = metodo_pagamento
        self.data_pagamento = data_pagamento
    
    def detalhes_pagamento(self):
        print(f"ID_Pagamento: {self.pagamento_id} | ID_Cliente: {self.cliente_id} | Valor Pago: R${self.valor_pago} | Metodo de Pagamento: {self.metodo_pagamento} | Data: {self.data_pagamento}")

compra1 = Pagamento(4090, 12345, 3690.00, "Pix", agora)

compra1.detalhes_pagamento()
```
