# Exercicio Funções

```python
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados tabulares 
import numpy as np  # Importa a biblioteca numpy para operações matemáticas e manipulação de arrays 

# Definindo a semente para gerar sempre os mesmos números aleatórios 
np.random.seed(27) 

# Criando um DataFrame com colunas ID, Nome e Idade 
df_pessoas = pd.DataFrame({ 
'ID': range(1, 31),  # Coluna ID com valores de 1 a 30 
'nome': ['Alice', 'Bruno', 'Carla', 'Daniel', 'Eduarda', 'Fernando', 
'Gabriela', 'Henrique', 'Isabela', 'João', 
'Karla', 'Lucas', 'Mariana', 'Nathan', 'Olivia', 'Paulo', 'Quezia', 
'Ricardo', 'Sabrina', 'Thiago', 
'Ursula', 'Vinicius', 'William', 'Xavier', 'Yasmin', 'Zaqueu', 
'Amanda', 'Bernardo', 'Camila', 'Diego' ],  # Coluna Nome com uma lista de 30 nomes fictícios 
'idade': np.random.randint(18, 60, 30)  # Coluna Idade com valores aleatórios entre 18 e 59 
}) 

# Cria um segundo DataFrame com informações de salário e departamento 
df_funcionarios = pd.DataFrame({ 
'ID': range(1, 31),  # Coluna ID com valores de 1 a 30 
'salario': np.random.randint(3000, 10000, 30),  # Coluna Salário com valores aleatórios entre 3000 e 9999 
'departamento': np.random.choice(['TI', 'RH', 'Financeiro', 'Marketing', 'Vendas'], 30)  # Coluna Departamento com escolha aleatória entre 5 opções 
}) 

# Criando o terceiro DataFrame com a média salarial por departamento 
# Calculando a média salarial por departamento 
df_departamentos = df_funcionarios.groupby('departamento')['salario'].mean().reset_index() 

# Renomeando a coluna para 'media_salario' 
df_departamentos.rename(columns={'salario': 'media_salario'}, inplace=True)

display(df_pessoas.head())
display(df_funcionarios.head())
display(df_departamentos.head())
```

# Ex 1
    Crie uma função que receba df_pessoas e retorne a média das idades. 

```python
def media_idades(df_pessoas):
    media = df_pessoas["idade"].sum() / len(df_pessoas)
    return media
print(f"A média das idades é: {media_idades(df_pessoas):.2f}")
```

# Ex 2
    Crie uma função que receba df_funcionarios e retorne o salário máximo e mínimo.

```python
def salario_maximo_minimo(df_funcionarios):
    maximo = df_funcionarios["salario"].max()
    minimo = df_funcionarios["salario"].min()
    return maximo, minimo

v_max, v_min = salario_maximo_minimo(df_funcionarios)
print(f"Salário máximo: R${v_max}, Salário mínimo: R${v_min}")
```

# Ex 3
    Escreva uma função que filtre df_pessoas para retornar apenas pessoas acima de 42 anos. 

```python
def filtrar_acima_42(df_pessoas):
    df_filtrado = df_pessoas[df_pessoas["idade"] > 42]
    return df_filtrado
resultado = filtrar_acima_42(df_pessoas)
display(resultado)
```

# Ex 4
    Crie uma função que una df_pessoas e df_funcionarios com base na coluna 'ID' 

```python
def uniao_ID(df_pessoas, df_funcionarios):
    df_unido_id = pd.merge(df_pessoas, df_funcionarios, on="ID")
    return df_unido_id
resultado = uniao_ID(df_pessoas, df_funcionarios)
display(resultado)
```

# Ex 5
    Escreva uma função que adicione uma nova linha ao df_pessoas.   

```python
def adicionar_pessoas(df_pessoas, ID , nome, idade):
    nova_info = {
        "ID": [ID],
        "nome": [nome],
        "idade": [idade]
    }
    df_nova_linha = pd.DataFrame(nova_info)
    df_atualizado = pd.concat([df_pessoas, df_nova_linha], ignore_index=True)
    return df_atualizado

df_pessoas = adicionar_pessoas(df_pessoas, 31, "João Luís", 22)
display(df_pessoas.tail())
```

# Ex 6

# Ex 7
    Escreva uma função que conte quantos funcionários há em cada departamento usando df_funcionarios. 

```python
def quantidade_por_departamento(df_funcionarios):
    quantidade = df_funcionarios.groupby("departamento")["ID"].count()
    return quantidade
resultado = quantidade_por_departamento(df_funcionarios)
display(resultado)
```

# Ex 8
Crie uma função que some os salários por departamento usando df_funcionarios. 

```python
def soma_salarios_departamento (df_funcionarios):
    soma = df_funcionarios.groupby("departamento")["salario"].sum().sort_values(ascending=False)
    return soma
resultado = soma_salarios_departamento(df_funcionarios)
display(resultado)
```

# Ex 9

# Ex 10
Faça uma função que receba df_pessoas e um nome, e retorne a idade correspondente. 

```python
def idade_correspondente_nome (df_pessoas, nome):
    idade = df_pessoas[df_pessoas["nome"] == nome]["idade"].values[0]
    return idade

nome_buscado = input("Digite um nome para ser encontrado:").title()
resultado_da_busca = idade_correspondente_nome(df_pessoas, nome_buscado)
print(f"A idade de {nome_buscado} é {resultado_da_busca} anos.")
```

# Ex 12
Faça uma função que retorne o nome do funcionário com maior salário, unindo df_pessoas e df_funcionarios.

```python
def funcionario_maior_salario(df_pessoas, df_funcionarios):
    df_merge = pd.merge(df_pessoas, df_funcionarios, on= "ID")
    linha_do_maior = df_merge["salario"].idxmax()
    maior_salario = df_merge.loc[linha_do_maior, "nome"].title()
    return maior_salario

resultado = funcionario_maior_salario(df_pessoas, df_funcionarios)
display(resultado)
```

# Ex 14
Crie uma função que compare os salários reais de df_funcionarios com as médias de df_departamentos e indique quais estão acima da média do setor.

```python
def comparacao_salario_media(df_funcionarios, df_departamentos):
    df_merge = pd.merge(df_funcionarios, df_departamentos, on="departamento")
    df_acima = df_merge[df_merge["salario"] > df_merge["media_salario"]]
    return df_acima
resultado = comparacao_salario_media(df_funcionarios, df_departamentos)
display(resultado)
```

# Ex 15
Faça uma função que crie uma nova coluna em df_funcionarios indicando se o salário está acima ou abaixo da média do setor (baseando-se no df_departamentos).

```python
def coluna_status(df_funcionarios, df_departamentos):
    df_merge = pd.merge(df_funcionarios, df_departamentos, on="departamento")

    df_merge["status"] = "Abaixo da Média"

    df_merge.loc[df_merge["salario"] == df_merge["media_salario"], "status"] = "Média"

    df_merge.loc[df_merge["salario"] > df_merge["media_salario"], "status"] = "Acima da Média"

    return df_merge
display(coluna_status(df_funcionarios, df_departamentos))
```
