## Introdução

Chamamos de Programação orientada a objetos (POO, como abreviação/ou OOP da sigla em inglês) um *paradigma* de programação no qual introduzimos objetos, que são instâncias de classes. Tais estruturas permitem ecapsular comportamentos, organizar o código de forma mais modular e utilizar alguns recursos especiais para extender comportamentos e reaproveitar códigos.

Mas antes, vamos entender o que é uma classe e o que é um objeto. Classe diz respeito ao *template*, ou seja é o modelo. Por outro lado, objeto é a classe instanciada, isso é uma classse em uso. 

Podemos fazer uma analogia com tipo de variáveis. 2 é um objeto da classe inteiro, "aristoteles" é um objeto da classe string (str), ...

Em python, uma classe é criada pela instrução class:

```py
class MinhaClasse():
    # código da classe 
```

Existem vários exemplos de classes com as quais já esbarramos em nossos estudos: `pd.DataFrame`, `np.array`, ...


## Atributos

Muitas vezes ao criar uma classe queremos definir algumas características delas, ou seja, propriedades que elas carregam.

Atributos podem ser criados de algumas formas diferentes. Em particular podemos definir ou modificar atributos de um objeto já criado, ou seta-los no momento em que *inicializamos* esse objeto. Exemplo:

```py
import pandas as pd
df = pd.DataFrame({"A":[1,2], "B":[3,4]})
df
```

A forma como definimos quais argumentos uma classe deve receber no momento que é inicializada é através de um método especial (`__init__`), que também é chamado de *construtor*. Vamos traduzir isso:

```py
class MinhaClasse():
    def __init__(self, arg1, arg2):
        pass
```

Na estrutura acima definimos **dentro** da classe `MinhaClasse` um método (que é similar a uma função). Nesse caso, esse método é um caso especial e por isso tem um nome fixo (`__init__`). Ele recebe um conjunto de argumentos, porém em especial ele precisa começar com o argumento `self`, que é uma forma de representar a própria classe. Com ele, podemos acessar informações e métodos da própria classe dentro da sua definição.

Além disso, temos outros argumentos `arg1, arg2` que podem aparecer em quantidade arbritrária. Podemos salvá-los como atributos da nossa classe da seguinte maneira:

```py
class MinhaClasse():
    def __init__(self, arg1, arg2):
        self.arg1 = arg1
        self.arg2 = arg2
```

Vamos traduzir o que o código acima faz:

1. Ele implementa uma classe chamada `MinhaClasse`,
2. Ela tem um método de inicialização com **dois** argumentos (`arg1, arg2`). Dessa forma seu uso no código seria `minha_casse = MinhaClasse(1, 2)`
3. Ao fazer essa inicialização os valores de 1 e 2 seriam salvos nos atributos `arg1` e `arg2`.

## Métodos

Atributos são características, ou seja, parâmetros de nosso objeto. Porém, muitas vezes objetos precisam realizar ações, que no caso de objetos são **métodos**. Acredite se quiser, mas já vimos vários métodos até agora. Por exemplo, se `df` é um dataframe do pandas:

```py
df.describe() # método que faz um resumo de estatísticas dos campos do dataframe
df.isna() # método que verifica se cada elemento de um dataframe está vazio
```

Você já deve ter reparado algo interessante:
1. Atributos são acessados com `.nome_do_atributo`
2. Métodos são acessados com `.nome_do_metodo()`

Sim, o parênteses distingue as duas categorias.

Mas vamos entender melhor:
| **Métodos** | **Atributos** |
|-------------|---------------|
| Executam ações ou comportamentos do objeto | Armazenam dados ou características do objeto |
| Podem receber argumentos | Não recebem argumentos diretamente, apenas armazenam valores |
| São definidos usando a palavra-chave `def` | São definidos diretamente na classe ou no construtor `__init__` |
| Precisam ser chamados para serem executados (ex: `obj.metodo()`) | São acessados diretamente (ex: `obj.atributo`) |
| Podem modificar os atributos do objeto | Guardam o estado atual do objeto |


Um exemplo de implementação de um método seria:

```py
class MinhaClasse():
    def __init__(self, arg1, arg2):
        self.arg1 = arg1
        self.arg2 = arg2
    def metodo_x(self, args):
        # faz alguma coisa
        return alguma_coisa # se quiser
```

Notem que podemos acessar o atributo. Nem sempre esse é um comportamento desejado, muitas vezes queremos proteger os atributos de um objeto, para fazer isso temos podemos usar modificadores de acesso, a serem usados como prefixo (antes do nome do atributo ou método):

1. _ (um underscore): é considerado uma proteção, ainda permite acesso mas serve para sinalizar que não deve acessar. 
2. __ (dois underscore): é utilizado para garantir privacidade, evita que um atributo ou método seja acessado de fora da classe (após sua construção).

## Decoradores

Decoradores em Python são uma maneira simples e poderosa de modificar o comportamento de funções ou métodos sem alterar o código original delas. Eles funcionam como "envoltórios" (ou "wrappers") que adicionam funcionalidades extras a uma função.

Exemplo:

```py
class MinhaClasse():
    def __init__(self, arg1, arg2):
        self.arg1 = arg1
        self.arg2 = arg2

    @decorador
    def metodo_x(self, args):
        # faz alguma coisa
        return alguma_coisa # se quiser
```

Nesse caso, o decorador `decorador` modifica o funcinamento do método `metodo_x`, ele pode adicionar um cronômetro, adicionar um log de início e fim, etc.

Um particularmente interessante é o `@property` ele serve para lidar com atriubos privados:

```py
class MinhaClasse():
    def __init__(self, arg1, arg2):
        self.__arg1 = arg1
        self.__arg2 = arg2

    @property
    def arg1(self):
        return self.__arg1

    @decorador
    def metodo_x(self, args):
        # faz alguma coisa
        return alguma_coisa # se quiser
```

Nesse código, criamos uma função com dois atributos privados (`__arg1` e `__arg2`). Depois criamos um método `arg1()` que retorna o valor do atributo. Com isso, podemos acessar o atributo `__arg1` pelo método `arg1`, mas pelo uso do decorador `@property` duas coisas especiais acontecem:
1. Esse método, em especial, acessamos *sem* os parênteses,
2. Esse método **protege o atributo**.

# Herança e Polimorfismo

Além das funcionalidades apresentadas até aqui sobre POO há várias outras que podem ser elencadas, mas em especial vamos destacar *herança* e *polimorfismo*. 

Considere o conceito a partir de um exemplo. Temos inicialmente uma classe `Produto`, porém um novo produto que podemos ter no nosso catálogo tem uma complexidade adicional. No caso de eletrodomésticos, além de registrar o nome e preço do produto, precisamos saber qual a tensão (popularmente chamada voltagem) do aparelho. Nesse caso, voltagem seria um novo *atributo* da classe, porém temos um conflito:
1. Nem todos os produtos têm *tensão* de operação, portanto se fizessemos uma alteração da classe produto para incluir *tensão* teríamos esse atributo presente em qualquer produto, eletrodoméstico ou não.
2. Se fizessemos uma nova classe teríamos que repetir a implementação de atributos como `nome` e `preco` além de métodos como `imprime_caracteristicas`.

Portanto nós chegamos no seguinte problema: será que conseguimos criar uma nova classe que extende o comportamento de uma inicial? A resposta é sim. 

Um exemplo genérico seria: 

```py
class ClasseMae():
    def __init__(self, arg1, arg2, ...):
        ... 


class ClasseFilha(ClasseMae):
    def __init__(self, arg1, arg2, ..., novo_arg1, novo_arg2, ...):
        super().__init__(arg1, arg2, ...)
        ...

```

Vamos traduzir o que está acontecendo acima:

1. Criamos uma classe chamada `ClasseMae`, que tem um método construtor `__init__` o qual receber um número arbitrário de argumentos: `arg1`, `arg2`, ...
2. Depois, criamos uma nova classe `ClasseFilha`, porém ao declaramos ela passaos entre parênteses o nome da classe mãe: `ClasseFilha(ClasseMae)`
    - **Esse é o primeiro ponto de atenção para herança: uma classe herda comportamentos de uma classe mãe quando ao criarmos ela usamos a sintaxe: `class ClasseFilha(ClasseMae):`**
3. Fazemos nosso método construtor `__init__` para a classe filha que recebe argumentos: `arg1`, `arg2`, ..., `novo_arg1`, `novo_arg2`, ... em que os primeiros argumentos (`arg1`, `arg2`, ...) são correspondentes aos argumentos da clase mãe. Por outro lado, `novo_arg1`, `novo_arg2`, ... são os novos argumentos que são introduzidos na classe filha. 
    - **Esse é o segundo ponto importante, para criar uma classe filha e configurar um construtor que aproveite a classe mãe, precisamos usar como argumentos todos os parâmetros: da classe mãe e os novos da classe filha**
4. Em seguida, para acessar os métodos da classe mãe (como o `__init__`) precisamos de uma função chamada `super()`, em seguida acessamos o método desejado.
    - **Essa é outra característica de classes _filha_: para acessar um método da classe mãe _dentro da classe_, precisamos usar `super()`**

Vamos usar isso para implementar o exemplo.

Note que ao usarmos herança não foi preciso repetir de novo a implementação dos atributos `nome` e `preco`. Mas e os métodos? Ainda consguimos acessar os métodos da clase mãe (como `imprime_caracteristicas`)?

Mas e se nosso interesse fosse alterar o método na classe filha. Como ficaria?

Note que as duas classes `Produto` e `Eletrodomestico` compartilham um mesmo método. Dessa forma, objetos de classes diferentes poder ser tratados da mesma maneira. Chamamos essa característica de *polimorfismo*.
