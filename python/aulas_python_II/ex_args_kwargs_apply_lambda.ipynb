# Exercicios *Args, **kwargs, Apply e Lambda

```python
import pandas as pd 
import numpy as np 

# Definindo a semente do gerador de números aleatórios para garantir reprodutibilidade 
np.random.seed(27) 

# Criando DataFrame 1 - Informações de vendas 
df_vendas = pd.DataFrame({ 
'ID_Venda': range(1, 31),  # Criando IDs únicos para cada venda (de 1 a 30) 
'Produto': np.random.choice(['Camisa', 'Calça', 'Tenis', 'Bone'], 30),  
# Produto vendido 
'Valor': np.random.uniform(50, 300, 30).round(2),  # Preço da venda 
'Quantidade': np.random.randint(1, 5, 30),  # Quantidade vendida 
'ID_Cliente': np.random.randint(1, 31, 30),  # Associando aleatoriamente um cliente à venda 
'ID_Funcionario': np.random.randint(1, 31, 30)  # Associando aleatoriamente um funcionário à venda 
}) 

# Lista de nomes para clientes 
nomes_clientes = [ 
'Lucas', 'Mariana', 'Carlos', 'Fernanda', 'Pedro', 'Ana', 'Rafael', 
'Juliana', 'Bruno', 'Camila', 
'Gustavo', 'Isabela', 'Thiago', 'Patricia', 'Felipe', 'Larissa', 
'Andre', 'Vanessa', 'Diego', 'Carla', 
'Ricardo', 'Natalia', 'Fabio', 'Bianca', 'Eduardo', 'Tatiane', 
'Leonardo', 'Aline', 'Joao', 'Leticia' ] 

# Criando DataFrame 2 - Informações de clientes 
df_clientes = pd.DataFrame({ 
'ID_Cliente': range(1, 31),  # Criando IDs únicos para os clientes (de 1 a 30) 
'Nome_Cliente': nomes_clientes,  # Nomes dos clientes 
'Idade': np.random.randint(18, 60, 30),  # Idade dos clientes 
'Cidade': np.random.choice(['Recife', 'Caruaru', 'Maceio', 'Joao Pessoa', 'Natal'], 30)  # Cidade dos clientes 
}) 

# Lista de nomes para funcionários 
nomes_funcionarios = [ 
'Roberto', 'Cristina', 'Fernando', 'Beatriz', 'Mateus', 'Sofia', 
'Daniel', 'Alessandra', 'Jose', 'Paula', 
'Antonio', 'Debora', 'Renato', 'Simone', 'Gabriel', 'Eliane', 'Cesar', 
'Adriana', 'Marcelo', 'Veronica', 
'Sergio', 'Regina', 'Alexandre', 'Claudia', 'Vinicius', 'Sandra', 
'Hugo', 'Renata', 'Paulo', 'Monica' ] 

# Criando DataFrame 3 - Informações de funcionários 
df_funcionarios = pd.DataFrame({ 
'ID_Funcionario': range(1, 31),  # Criando IDs únicos para os funcionários (de 1 a 30) 
'Nome_Funcionario': nomes_funcionarios,  # Nomes dos funcionários 
'Cargo': np.random.choice(['Vendedor', 'Gerente', 'Caixa', 'Repositor'], 30),  # Cargo dos funcionários 
'Salario': np.random.uniform(1500, 5000, 30).round(2)  # Salário dos funcionários 
})
```

# Ex 1
Use ´apply´ para calcular o valor total de cada venda no DataFrame ‘df_vendas’ (Valor * Quantidade).

```python
def total_venda(linha):
    calculo = linha["Valor"] * linha["Quantidade"]
    return calculo
df_vendas["Total"] = df_vendas.apply(total_venda,axis=1)
display(df_vendas.head())
```

# Ex 2
Crie uma função que utilize lambda para formatar os nomes dos clientes do DataFrame ‘df_clientes’ para maiúsculas.

```python
df_clientes["Nome_Cliente"] = df_clientes["Nome_Cliente"].apply(lambda x: x.upper())
display(df_clientes)
```

# Ex 3
Utilize apply para categorizar os clientes do DataFrame ‘df_clientes’ em ‘Jovem’ (idade < 30) e ‘Adulto’ (idade >= 30).

```python
def define_faixa_etaria(Idade):
    if Idade < 30:
        faixa_etaria = "Jovem"
    elif Idade >= 30:
        faixa_etaria = "Adulto"
    return faixa_etaria
df_clientes["faixa_etaria"] = df_clientes["Idade"].apply(define_faixa_etaria)
display(df_clientes)
```

```python
#opção com lambda
df_clientes["faixa_etaria"] = df_clientes["Idade"].apply(lambda x: "Jovem" if x < 30 else "Adulto")
display(df_clientes)
```

# Ex 4
Use apply para converter os salários dos funcionários do DataFrame ‘df_funcionarios’ de Reais para Dólares (cotação fictícia).

```python
df_funcionarios["Salario"] = df_funcionarios["Salario"].apply(lambda x: x / 5.21).round(2)
display(df_funcionarios.head())
```

# Ex 5
Crie uma função que utilize *args para calcular a média dos valores da coluna ‘Valor’ do DataFrame ‘df_vendas’.

```python
def media_valor(*args):
    soma = sum(args)
    quantidade = len(args)
    return soma / quantidade
resultado = media_valor(*df_vendas["Valor"])
print(f"a Média dos valores é: {resultado:.2f}")
```

# Ex 6
Utilize **kwargs para criar uma função que filtre produtos específicos no DataFrame ‘df_vendas’.

```python
def filtro_produtos(**kwargs):
    df_filtrado = df_vendas.copy()

    for coluna, valor in kwargs.items():
        df_filtrado = df_filtrado[df_filtrado[coluna] == valor]
    return df_filtrado
resultado1 = filtro_produtos(Produto="Camisa")
display(resultado1)
resultado2 = filtro_produtos(Produto = "Camisa", Quantidade = 2)
display(resultado2)
```
